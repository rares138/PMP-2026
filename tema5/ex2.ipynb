```python
def myViterbi(pi, A, B, sir):
  N = len(pi)
  T = len(sir)
  delta = np.zeros((T, N))
  psi = np.zeros((T,N), dtype=int)

  delta[0] = pi * B[:, sir[0]]

  for t in range(1, T):
      for j in range(N):
          options = delta[t-1]*A[:,j]
          delta[t,j] = B[j, sir[t]]*np.max(options)
          psi[t,j] = np.argmax(options)
  return delta, psi
```

Ψ_t(j) retine starea anterioara i care maximizeaza calculul lui δ_t(j), adica cel mai bun predecesor al starii j la pasul t. La final, alegem starea finala optima si reconstruim traseul prin backtracking folosind Ψ_t.
